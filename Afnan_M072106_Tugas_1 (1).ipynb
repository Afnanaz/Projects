#Importig library

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
```

# Creating dataset

```python
# Data teks
texts = [
    "statistika susah banget! ga cuman mean aja",
    "dosennya keren, cara ngajarnya mudah dipahami",
    "Kursinya rusak-rusak, ga nyaman",
    "ruangannya sejuk",
    "fasilitas labnya memadai"
]
labels = ["Negatif", "Positif", "Negatif", "Positif", "Positif"]
```

# Converting dataset into Numerical form

```python
# Mengubah label menjadi numerik
label_map = {"Negatif": 0, "Positif": 1}
numerical_labels = [label_map[label] for label in labels]
```

# Bag of Word

```python
# Membuat dan menerapkan CountVectorizer (Bag of Words)
count_vectorizer = CountVectorizer()
X_bow = count_vectorizer.fit_transform(texts)
```

```python
# Menampilkan kosakata dan representasi BoW
vocab_bow = count_vectorizer.get_feature_names_out()
bow_array = X_bow.toarray()
print("Vocabulary:", vocab_bow)
print("Bag of Words Representation:\n", bow_array)
```

# TF-IDF

```python
# Membuat dan menerapkan TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()
X_tfidf = tfidf_vectorizer.fit_transform(texts)
```

```python
# Konversi hasil TF-IDF ke dalam DataFrame untuk visualisasi
words_tfidf = tfidf_vectorizer.get_feature_names_out()
df_tfidf = pd.DataFrame(X_tfidf.toarray(), columns=words_tfidf)
print(df_tfidf)
```

# Spitting Dataset

```python
# Memisahkan data untuk pelatihan dan pengujian
X_train_bow, X_test_bow, y_train, y_test = train_test_split(bow_array, numerical_labels, test_size=0.2, random_state=42)
X_train_tfidf, X_test_tfidf = train_test_split(df_tfidf, test_size=0.2, random_state=42)
```

# Modelling

```python
# RandomForest dengan Bag of Words
rf_bow = RandomForestClassifier(random_state=42)
rf_bow.fit(X_train_bow, y_train)
y_pred_bow = rf_bow.predict(X_test_bow)
print("Accuracy:", accuracy_score(y_test, y_pred_bow))
print(classification_report(y_test, y_pred_bow, labels=[0, 1], target_names=["Negatif", "Positif"]))
```

```python
# RandomForest dengan TF-IDF
rf_tfidf = RandomForestClassifier(random_state=42)
rf_tfidf.fit(X_train_tfidf, y_train)
y_pred_tfidf = rf_tfidf.predict(X_test_tfidf)

print("Accuracy:", accuracy_score(y_test, y_pred_tfidf))
print(classification_report(y_test, y_pred_tfidf, labels=[0, 1], target_names=["Negatif", "Positif"]))
```
